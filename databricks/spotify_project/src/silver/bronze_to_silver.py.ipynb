```python
from pyspark.sql.functions import current_timestamp



"""
schemaEvolutionMode
addNewColumns — add new columns to inferred schema; stream stops and needs restart
rescue — do not evolve schema; unexpected fields go to _rescued_data
failOnNewColumns — fail when new columns appear
none — ignore new columns; no rescue unless rescuedDataColumn is set
addNewColumnsWithTypeWidening — adds new columns and widens supported types

cloudFiles.schemaEvolutionMode = file ingestion behavior
mergeSchema / overwriteSchema = Delta table write behavior

"""
#Define isolated metadata directories (Must be unique and separate folders)
schema_checkpoint_path = "abfss://silver@manasmandal14storage.dfs.core.windows.net/DimUser/schema"
write_checkpoint_path  = "abfss://silver@manasmandal14storage.dfs.core.windows.net/DimUser/sink"
target_silver_path = "abfss://silver@manasmandal14storage.dfs.core.windows.net/DimUser/data"

#Build the Streaming DataFrame Reader using Auto Loader (cloudFiles)
df_streaming_source = (spark.readStream
  .format("cloudFiles")
  .option("cloudFiles.format", "parquet")
  .option("cloudFiles.schemaLocation", schema_checkpoint_path)
  .option("cloudFiles.schemaEvolutionMode", "addNewColumns")
  .option("cloudFiles.includeExistingFiles", "true")  # Forces read of files already sitting in bronze
  .load("abfss://bronze@manasmandal14storage.dfs.core.windows.net/DimUser"))


# 4. Write the stream directly into your Target Managed Table
query = (df_streaming_source.writeStream
  .format("delta")
  .option("checkpointLocation", write_checkpoint_path)
  .outputMode("append")
  .option("mergeSchema", "true")
  .trigger(availableNow=True)  # Processes new data incrementally as a batch, then stops the stream
  .option("path","abfss://silver@manasmandal14storage.dfs.core.windows.net/DimUser/data")
  .toTable("spotify_catalogue.dev_schema.DimUser"))


```
